```python
import nltk
import matplotlib
import tkinter
```

```python
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
text='I saw a dog roaming on the street'
```

```python
tokens=nltk.word_tokenize(text)
print(tokens)
```

```python
tag=nltk.pos_tag(tokens)
print("\n")
print(tag)
```

```python
pattern=""" 
NP:{<RP>?<JJ>*<LS>}
IN:{<IN>}
VBD:{<VBD>}
"""

np_chunk=nltk.RegexpParser(pattern)
res=np_chunk.parse(tag)
```

```python
res.draw()
```
