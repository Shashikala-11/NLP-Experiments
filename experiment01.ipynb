```python
while True:
    try:
        age=int(input("How old you are?"))
        break
    except ValueError:
        print("Please enter a whole no.:")
print("Your age is"+str(age))
    
```

```python
validInt=False
while not validInt:
    age=input("How old are you?")
    if age.isdigit():
        validInt=True
    else:
        print("You must enter a valid number")
print("Your age is"+str(age))        
```

```python
isLong=False
while not isLong:
    password=input("Enter password of atleast 5 char:")
    if len(password)>=5:
        isLong=True
    if not isLong:
        print('Your password entered should be atleast of min 5 char '+password)
    else:
         print('Your password: '+password)
    
```
