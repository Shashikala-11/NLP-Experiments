### Experiment 02: To apply various other text preprocessing techniques for any given text: Tokenization and Stop Word Removal

### 1. Stop words removal program

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
print(stopwords.words('english'))
```

```python
sentence='Hi , This is Shashikala Gupta . I am a Data Science Student.'
```

```python
stop_words=set(stopwords.words('english'))
word_tokens=word_tokenize(sentence)
filtered_sentence=[]
```

```python
for w in word_tokens:
    if w not in stop_words:
        filtered_sentence.append(w)
print('\n')
print(word_tokens)
print(filtered_sentence)        

```

### 2. Sentence Tokenization and word tokenization

```python
from nltk.tokenize import sent_tokenize
```

```python
text='Hiii. Bye'
t=sent_tokenize(text)
print('Sentence tokenization:')
for i in t:
    print(f'Sentence-->{i}')
```
