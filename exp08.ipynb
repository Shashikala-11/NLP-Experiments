#### Program 02

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize,sent_tokenize
stop_words=set(stopwords.words('english'))
```

```python
txt=''' The internet has changed the way people communicate. It allows instant messaging, video calls, and easy sharing of information across the world. Because of these advantages, people can connect quickly and stay informed'''
tokenized=sent_tokenize(txt)
```

```python
for i in tokenized:
    word_list=nltk.word_tokenize(i)
    tagged=nltk.pos_tag(word_list)

    print(tagged)
```
