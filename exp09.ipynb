#### Experiment 09 :- To identify and segment the named entities and classify or categorize them under various predefined classes.

```python
import nltk
from nltk import word_tokenize,pos_tag
```

```python


text="NASA awarded Elon Musk's SpaceX a $2.9 billion contract to build the lunar lander."
tokens=word_tokenize(text)
tag=pos_tag(tokens)
print(tag)

ne_tree=nltk.ne_chunk(tag)
print(ne_tree)
```
