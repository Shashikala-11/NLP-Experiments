#### Experiment 10:- To extract information from text using Chunking methods

```python
import nltk
nltk.download('brown')

from nltk import pos_tag
from nltk import word_tokenize
from nltk import RegexpParser

text=''' The internet has changed the way people communicate. It allows instant messaging, video calls, and easy sharing of information across the world. Because of these advantages, people can connect quickly and stay informed'''
```

```python
list_of_words=word_tokenize(text)
tagged_words=pos_tag(list_of_words)
chuncks_to_be_extracted='''Chunk:{<DT>*<NNP>*<NN>}'''
```

```python
chunkParser=nltk.chunk.RegexpParser(chuncks_to_be_extracted)
chunked_sentence=chunkParser.parse(tagged_words)
```

```python
chunked_sentence.draw()
```

```python
print('chunks obtained:\n')
for subtree in chunked_sentence.subtrees():
    if subtree.label()=='Chunk':
        print(subtree)
```
