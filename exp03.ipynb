### Experiment 03: To apply various other text preprocessing techniques for any given text : Stemming

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
```

```python
ps=PorterStemmer()
```

```python
words=['engine','engineer','engineering','engineered']
for i in words:
    print(i,"-->",ps.stem(i))
```

#### stemming words from sentences

```python
sentence='hello shashi....Did you eat meal before meaningful work worked by you in indoor.'
words=word_tokenize(sentence)

for i in words:
    print(i,"-->",ps.stem(i))
```
